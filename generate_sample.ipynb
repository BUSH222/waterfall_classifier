```python
import os
import random
import shutil


indir = 'data'
outdir = 'data_sample'
withsignal = 'with_signal'
withoutsignal = 'without_signal'
num_samples = 1500
```

```python
os.makedirs(os.path.join(outdir, withsignal), exist_ok=True)
os.makedirs(os.path.join(outdir, withoutsignal), exist_ok=True)

sample_size = num_samples // 2

with_signal_files = random.sample(os.listdir(os.path.join(indir, withsignal)), sample_size)
for file in with_signal_files:
    src = os.path.join(indir, withsignal, file)
    dst = os.path.join(outdir, withsignal, file)
    shutil.copy2(src, dst)

without_signal_files = random.sample(os.listdir(os.path.join(indir, withoutsignal)), sample_size)
for file in without_signal_files:
    src = os.path.join(indir, withoutsignal, file)
    dst = os.path.join(outdir, withoutsignal, file)
    shutil.copy2(src, dst)

print(f"Copied {sample_size} files from {withsignal} and {sample_size} files from {withoutsignal}")
```
